```python
class Greeter(object):
    def __init__(self,name):
        self.name=name
    
    def greet(self, loud=False):
        if loud:
            print('HELLO, %s!'%self.name.upper())
        else:
            print('Hello, %s' % self.name)

g=Greeter('Fred')
g.greet()
g.greet(loud=True)
```

broadcasting：
维度对齐：如果两个数组的维度数量（秩）不同，NumPy 会在维度较少的数组的形状前面补 1，直到它们的维度数量相同。

例如：一个形状为 (4, 3) 的数组和一个形状为 (3,) 的数组相加。后者形状会先被视为 (1, 3)。
兼容性检查：在维度对齐后，从末尾维度开始逐个比较它们的尺寸大小。在任何一个维度上，如果满足以下两个条件之一，则称它们是兼容的：

两个数组在该维度上的尺寸相等。
其中一个数组在该维度上的尺寸为 1。 如果所有维度都兼容，则可以进行广播。否则，会引发 ValueError。
结果形状：广播后，结果数组的形状是两个输入数组在各个维度上尺寸的最大值。

“拉伸”行为：在计算时，尺寸为 1 的维度会被“拉伸”或“复制”，以匹配另一个数组在该维度上的尺寸，但这个过程并不会在内存中实际创建副本。

```python
v=np.array([1,2,3])
w=np.array([4,5])

print(np.reshape(v,(3,1))*w)

x=np.array([[1,2,3],[4,5,6]])
print(x+v)

print((x.T + w).T)

print(x + np.reshape(w, (2, 1)))
print(x * 2)


```

```python
from scipy.misc import imread, imsave, imresize

# Read an JPEG image into a numpy array
img = imread('assets/cat.jpg')
print(img.dtype, img.shape)  # Prints "uint8 (400, 248, 3)"

# We can tint the image by scaling each of the color channels
# by a different scalar constant. The image has shape (400, 248, 3);
# we multiply it by the array [1, 0.95, 0.9] of shape (3,);
# numpy broadcasting means that this leaves the red channel unchanged,
# and multiplies the green and blue channels by 0.95 and 0.9
# respectively.
img_tinted = img * [1, 0.95, 0.9]

# Resize the tinted image to be 300 by 300 pixels.
img_tinted = imresize(img_tinted, (300, 300))

# Write the tinted image back to disk
imsave('assets/cat_tinted.jpg', img_tinted)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0,3*np.pi,0.1) #0到pi/3,步长为0.1。一个numpy数组，[0.,0.1,0.2,....,9.4]
y=np.sin(x)

plt.plot(x,y)
plt.show()
```

```python
y_sin=np.sin(x)
y_cos=np.cos(x)


plt.plot(x,y_sin)
plt.plot(x,y_cos)
plt.xlabel('x axix')
plt.ylabel('y axis')
plt.title('Sine and Cosine')
plt.legend(['Sine','Cosine']) #按照plot调用顺序
plt.show()

```

```python
plt.subplot(2,1,1)  #垂直划分为2行。水平划分为1列，2*1的网格里的第一个子图
plt.plot(x,y_sin)
plt.title('Sine')

plt.subplot(2,1,2)
plt.plot(x,y_cos)
plt.title('Cosine')

plt.show()
```

```python
import numpy as np
from scipy.misc import imread, imresize
import matplotlib.pyplot as plt

img = imread('assets/cat.jpg')
img_tinted = img * [1, 0.95, 0.9]

# Show the original image
plt.subplot(1, 2, 1)
plt.imshow(img)

# Show the tinted image
plt.subplot(1, 2, 2)

# A slight gotcha with imshow is that it might give strange results
# if presented with data that is not uint8. To work around this, we
# explicitly cast the image to uint8 before displaying it.
plt.imshow(np.uint8(img_tinted))
plt.show()
```
